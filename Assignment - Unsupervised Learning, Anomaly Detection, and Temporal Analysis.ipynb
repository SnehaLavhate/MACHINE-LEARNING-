##### Question 1 :What is Dimensionality Reduction? Why is it important in machine learning?

Dimensionality Reduction is the process of reducing the number of input variables (features) while preserving as much relevant information as possible.

Importance:

* Reduces computational cost

* Removes noise and redundant features

* Helps visualization (2D/3D)

* Improves model performance and reduces overfitting

##### Question 2: Name and briefly describe three common dimensionality reduction techniques..

1. PCA (Principal Component Analysis) – Transforms data into new components that capture maximum variance.

2. t-SNE (t-distributed Stochastic Neighbor Embedding) – Non-linear technique useful for visualizing high-dimensional data in 2D/3D.

3. LDA (Linear Discriminant Analysis) – Supervised method that finds feature combinations maximizing class separation.

##### Question 3: What is clustering in unsupervised learning? Mention three popular clustering algorithms

Clustering is an unsupervised learning technique that groups similar data points based on patterns or similarity, without predefined labels.

Popular clustering algorithms:

1. K-Means

2. Hierarchical Clustering

3. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

##### Question 4: Explain the concept of anomaly detection and its significance.

Anomaly Detection identifies rare events, outliers, or unusual patterns that deviate from normal behavior.

Significance:

* Helps detect fraud (banking/transactions)

* Identifies system failures or faults

* Finds unusual behavior in network security

* Useful in healthcare monitoring and sensor data analysis

##### Question 5: List and briefly describe three types of anomaly detection techniques

1. Statistical Methods – Use probability distributions, z-scores, thresholds to detect deviations.

2. Machine Learning Methods – Algorithms like Isolation Forest, One-Class SVM detect outliers.

3. Distance-Based Methods – Use k-NN distance or clustering to identify points far from neighbors.

##### Question 6: What is time series analysis? Mention two key components of time series data.

Time Series Analysis is the study of data recorded over time to analyze trends, patterns, and forecasting future values.

Key components:

1. Trend – Long-term upward or downward movement.

2. Seasonality – Repeating patterns at regular intervals (daily/weekly/yearly).


##### Question 7: Describe the difference between seasonality and cyclic behavior in time series.


* Seasonality:

   - Occurs at fixed, regular intervals

   - Example: Increased ice cream sales every summer

* Cyclic Behavior:

    - Occurs at irregular, non-fixed intervals

    - Example: Economic recessions and expansions

```python
## Question 8: Write Python code to perform K-means clustering on a sample dataset.

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Create sample dataset
X, y = make_blobs(n_samples=300, centers=3, random_state=42)
```

```python
# Apply K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
labels = kmeans.labels_
```

```python
# Plot clusters
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=200, marker='X')
plt.show()
```

```python
## Question 9: What is inheritance in OOP? Provide a simple example in Python.

# Inheritance allows a class (child/subclass) to acquire the properties and methods of another class (parent/superclass), promoting code reuse.
class Animal:
    def sound(self):
        return "Some sound"

class Dog(Animal):   # Dog inherits from Animal
    def sound(self):
        return "Bark"

d = Dog()
print(d.sound())  
```

##### Question 10: How can time series analysis be used for anomaly detection?

Time series analysis detects anomalies by identifying time points where the value deviates significantly from the expected pattern.

Methods include:

* Forecasting models like ARIMA, LSTM to detect deviations between predicted & actual values

* Seasonal decomposition to find irregular components

* Statistical thresholds on moving averages
