## Question 1 : What is the fundamental idea behind ensemble techniques? How does bagging differ from boosting in terms of approach and objective? 
**Ans:**
**Fundamental Idea Behind Ensemble Techniques**
1. Ensemble techniques combine multiple weak learners (individual models) to create a strong learner.
2. The goal is to improve accuracy, stability, and generalization of predictions.
3. They help to reduce errors, variance, and bias in the model.
4. Final prediction is made by aggregating the outputs of all models (e.g., voting or averaging).

### Difference Between Bagging and Boosting
1. Approach
- Bagging:
  - Models are trained independently and in parallel.
  - Each model gets a random subset of data (sampled with replacement).

- Boosting:
  - Models are trained sequentially.
  - Each new model focuses on correcting the errors of the previous models.

2. Objective
- Bagging: To reduce variance and prevent overfitting.
- Boosting: To reduce bias and improve model accuracy.

3. Data Sampling
- Bagging: Uses bootstrap sampling (random samples with replacement).
- Boosting: Uses the entire dataset, but adjusts weights of samples based on previous errors.

4. Model Combination
- Bagging: Combines results using averaging (regression) or majority voting
(classification).
- Boosting: Combines results using weighted voting or weighted sum (more weight to better models).

5. Examples
- Bagging: Random Forest
- Boosting: AdaBoost, Gradient Boosting, XGBoost

## Question 2: Explain how the Random Forest Classifier reduces overfitting compared to a single decision tree. Mention the role of two key hyperparameters in this process.
**Ans:**
### 1. Reduction of Overfitting
1. A single decision tree often overfits the training data because it learns all details, including noise and outliers.
2. Random Forest reduces overfitting by building multiple decision trees and combining their results.
3. It uses the concept of bagging (Bootstrap Aggregating) — each tree is trained on a different random subset of the data.
4. It also introduces feature randomness — at each split, only a random subset of features is considered.
5. These two types of randomness ensure that trees are less correlated, which reduces overfitting and increases generalization.
6. The final prediction is made by averaging (for regression) or majority voting (for classification) across all trees, making results more stable and less sensitive to noise.

### 2. Key Hyperparameters and Their Role
- a) n_estimators
  - This hyperparameter defines the number of trees in the forest.
  - More trees lead to better averaging and lower variance, which helps in reducing overfitting.
  - However, too many trees can increase training time, so it must be chosen carefully.

- b) max_features
  - It specifies the number of features to consider when looking for the best split at each node.
  - By limiting the number of features per tree, Random Forest ensures that each tree is diverse, reducing correlation among trees.
  - This randomness further helps to avoid overfitting and improves generalization.

## Question 3: What is Stacking in ensemble learning? How does it differ from traditional bagging/boosting methods? Provide a simple example use case.
**Ans:**
1. Stacking (Stacked Generalization) is an ensemble learning technique that combines predictions from multiple different models (called base learners) using a meta-model (or meta-learner) to improve overall performance.
2. Instead of simply averaging or voting (like in bagging), stacking learns how to best combine the outputs of different models.
3. The base models can be of different types — e.g., decision trees, logistic regression, SVM, etc.
4. The meta-model is trained on the predictions of these base models (not the original features) to learn which model performs best in different situations.
5. This approach helps to capture the strengths of multiple algorithms, reducing both bias and variance.

### 2. Difference Between Stacking and Bagging/Boosting
1. Training Approach
  - Bagging: Trains multiple models of the same type independently on different random subsets of data.
  - Boosting: Trains models sequentially, each correcting the errors of the previous one.
  - Stacking: Trains different types of models in parallel, then combines their outputs using a meta-learner.

2. Combination Method
  - Bagging: Combines predictions by averaging or voting.
  - Boosting: Combines predictions using a weighted sum based on model accuracy.
  - Stacking: Combines predictions using a learned meta-model that decides how to weight or use each model’s output.

3. Goal
  - Bagging: Reduces variance.
  - Boosting: Reduces bias.
  - Stacking: Aims to improve predictive power by leveraging the strengths of diverse models.

4. Model Diversity
  - Bagging/Boosting: Usually use the same algorithm type (e.g., many decision trees).
  - Stacking: Uses different types of algorithms (e.g., logistic regression + decision tree + SVM).

3. Simple Example Use Case
- Suppose you are building a model to predict whether a loan will be approved or not.
- You train three base models:
- 1. Logistic Regression – good with linear relationships.
- 2. Decision Tree – good with non-linear patterns.
- 3. Support Vector Machine (SVM) – good with complex boundaries.
- Then, you train a meta-model (e.g., Random Forest or Linear Regression) on the outputs (predictions) of these three models.
- The meta-model learns how to combine their predictions to achieve better overall accuracy than any single model.

## Question 4:What is the OOB Score in Random Forest, and why is it useful? How does it help in model evaluation without a separate validation set?
**Ans:**
1. Meaning of OOB Score
- OOB (Out-of-Bag) Score stands for Out-of-Bag Score, which is an internal validation accuracy estimate used in Random Forest.
- In Random Forest, each tree is trained using a bootstrap sample — a random sample taken with replacement from the original training data.

- As a result, about one-third of the data is not included in the bootstrap sample for each tree.
- These unused samples are called Out-of-Bag (OOB) samples.
- The OOB Score is calculated by testing each tree on its own OOB samples and averaging the prediction accuracy across all trees.

2. Why OOB Score is Useful
- It provides a built-in cross-validation method during the training of the Random Forest.
- It helps to evaluate model performance without needing a separate validation or test set.
- This saves time and data, especially when the dataset is small.
- It helps to detect overfitting or underfitting early, as it gives an unbiased estimate of model accuracy.
- The OOB Score is often close to the test accuracy, making it a reliable performance indicator.

3. How It Helps in Model Evaluation (Without Separate Validation Set)
- Since every tree is tested on its own unseen OOB data, the OOB Score acts as a validation accuracy for the entire forest.
- The model effectively performs cross-validation internally during training.
- Therefore, there is no need to split the dataset into separate training and validation sets — the same data is used efficiently for both.
- This allows for better data utilization, especially when working with limited data.

## Question 5: Compare AdaBoost and Gradient Boosting in terms of:
- How they handle errors from weak learners
- Weight adjustment mechanism
- Typical use cases
  
**Ans:**
  
**1. How They Handle Errors from Weak Learners**
### AdaBoost:

  1. AdaBoost handles errors by assigning higher weights to the samples that were misclassified by previous weak learners.
  2. Each new weak learner focuses more on the hard-to-classify examples, trying to correct the mistakes made earlier.
  3. It sequentially improves performance by paying more attention to errors in classification.

- Gradient Boosting:

  1. Gradient Boosting handles errors by minimizing a loss function using gradient descent.
  2. Instead of reweighting samples, it builds new models that predict the residual errors (differences between actual and predicted values).
  3. Each new model is added to reduce the overall prediction error step by step.

**2. Weight Adjustment Mechanism**

### AdaBoost:
  1. Assigns weights to each training sample.
  2. Misclassified samples get increased weights, and correctly classified samples get decreased weights.
  3. Final prediction is based on a weighted majority vote (for classification) or weighted average (for regression).

- Gradient Boosting:
1. Uses gradients (slopes of the loss function) to decide how much correction is needed for each prediction.
2. Weights are not explicitly assigned to samples; instead, the model learns from residuals to improve step by step.
3. The learning rate (shrinkage parameter) controls how much each new tree contributes to the final prediction.

## 3. Typical Use Cases

**AdaBoost:**
1. Best suited for classification problems, especially binary classification.
2. Commonly used with Decision Stumps (one-level decision trees) as weak learners.
3. Useful when data is less noisy and features are well-separated.

- Gradient Boosting:
1. Suitable for both regression and classification problems.
2. Works well on complex, high-dimensional datasets.
3. Commonly used in real-world applications like ranking, forecasting, and structured data prediction (e.g., XGBoost, LightGBM).

## Question 6:Why does CatBoost perform well on categorical features without requiring extensive preprocessing? Briefly explain its handling of categorical variables.
**Ans:**
1. Key Reason for CatBoost’s Good Performance
- CatBoost (short for Categorical Boosting) is specifically designed to handle categorical features efficiently and automatically.
- Unlike other algorithms (like XGBoost or LightGBM), CatBoost does not require manual preprocessing such as one-hot encoding or label encoding.
- It internally converts categorical values into numerical representations in a way that preserves the natural relationships between categories and avoids overfitting.

2. How CatBoost Handles Categorical Variables
- 1. Target Statistics (Mean Encoding):
- CatBoost replaces categorical values with target-based statistics (like the average label value) instead of arbitrary numeric codes.
- For example, if the feature is “City,” CatBoost might replace each city with the average target value (e.g., average income for that city).
- 2. Permutation-Driven Encoding (to avoid target leakage):
- CatBoost uses a permutation-based approach to compute these statistics.
- For each sample, it calculates the target statistic using only the previous samples in a random order (permutation).
- This prevents target leakage (using information from the current or future samples) and ensures fair, unbiased encoding.
- 3. Efficient Combination of Categorical Features:
- CatBoost automatically creates combinations of categorical features (feature interactions) during training.
- This allows the model to capture complex relationships between multiple categorical variables.

3. Why Extensive Preprocessing is Not Required
- No need for manual one-hot encoding — saves memory and computation time.
- Automatically handles high-cardinality features (features with many unique values).
- Reduces the risk of overfitting that can occur with traditional target encoding.
- Produces fast, accurate, and stable models without manual tuning or transformation of categorical data.

## Question 7: KNN Classifier Assignment: Wine Dataset Analysis withOptimization
- Task:
1. Load the Wine dataset (sklearn.datasets.load_wine()).
2. Split data into 70% train and 30% test.
3. Train a KNN classifier (default K=5) without scaling and evaluate using:
a. Accuracy
b. Precision, Recall, F1-Score (print classification report)
4. Apply StandardScaler, retrain KNN, and compare metrics.
5. Use GridSearchCV to find the best K (test K=1 to 20) and distance metric
(Euclidean, Manhattan).
6. Train the optimized KNN and compare results with the unscaled/scaled versions.
**Ans:**

```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1️⃣ Load the Wine dataset
wine = load_wine()
X = wine.data
y = wine.target

# 2️⃣ Split data into 70% train and 30% test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

# 3️⃣ Train KNN classifier (default K=5) without scaling
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

print("\n--- Without Scaling ---")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# 4️⃣ Apply StandardScaler and retrain KNN
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn_scaled = KNeighborsClassifier(n_neighbors=5)
knn_scaled.fit(X_train_scaled, y_train)
y_pred_scaled = knn_scaled.predict(X_test_scaled)

print("\n--- With StandardScaler ---")
print("Accuracy:", accuracy_score(y_test, y_pred_scaled))
print("\nClassification Report:\n", classification_report(y_test, y_pred_scaled))

# 5️⃣ Use GridSearchCV to find best K (1–20) and metric (Euclidean, Manhattan)
param_grid = {
    'n_neighbors': list(range(1, 21)),
    'metric': ['euclidean', 'manhattan']
}

grid_search = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train_scaled, y_train)

print("\n--- GridSearchCV Results ---")
print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation Accuracy:", grid_search.best_score_)

# 6️⃣ Train optimized KNN and compare results
best_knn = grid_search.best_estimator_
y_pred_best = best_knn.predict(X_test_scaled)

print("\n--- Optimized KNN Model ---")
print("Accuracy:", accuracy_score(y_test, y_pred_best))
print("\nClassification Report:\n", classification_report(y_test, y_pred_best))
```

## Question 8 : PCA + KNN with Variance Analysis and Visualization
- Task:
1. Load the Breast Cancer dataset (sklearn.datasets.load_breast_cancer()).
2. Apply PCA and plot the scree plot (explained variance ratio).
3. Retain 95% variance and transform the dataset.
4. Train KNN on the original data and PCA-transformed data, then compare
accuracy.
5. Visualize the first two principal components using a scatter plot (color by class).
**Ans:**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1️⃣ Load the Breast Cancer dataset
data = load_breast_cancer()
X = data.data
y = data.target

print("Dataset shape:", X.shape)
print("Number of classes:", len(np.unique(y)))

# 2️⃣ Apply PCA and plot Scree Plot (explained variance ratio)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca_full = PCA()
pca_full.fit(X_scaled)

explained_variance = np.cumsum(pca_full.explained_variance_ratio_)

plt.figure(figsize=(8,5))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker='o')
plt.title("Scree Plot: Cumulative Explained Variance by PCA Components")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid(True)
plt.show()

# 3️⃣ Retain 95% variance and transform dataset
pca_95 = PCA(n_components=0.95)
X_pca = pca_95.fit_transform(X_scaled)

print(f"\nNumber of components to retain 95% variance: {pca_95.n_components_}")

# 4️⃣ Train KNN on original data and PCA-transformed data, compare accuracy

# Split original data
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42, stratify=y)

# Train on original data
knn_orig = KNeighborsClassifier(n_neighbors=5)
knn_orig.fit(X_train, y_train)
y_pred_orig = knn_orig.predict(X_test)
acc_orig = accuracy_score(y_test, y_pred_orig)

# Split PCA data
X_pca_train, X_pca_test, y_pca_train, y_pca_test = train_test_split(X_pca, y, test_size=0.3, random_state=42, stratify=y)

# Train on PCA-transformed data
knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(X_pca_train, y_pca_train)
y_pred_pca = knn_pca.predict(X_pca_test)
acc_pca = accuracy_score(y_pca_test, y_pred_pca)

print("\n--- Accuracy Comparison ---")
print(f"KNN on Original Data: {acc_orig:.4f}")
print(f"KNN on PCA Data (95% variance): {acc_pca:.4f}")

# 5️⃣ Visualize the first two principal components using a scatter plot
pca_2 = PCA(n_components=2)
X_2d = pca_2.fit_transform(X_scaled)

plt.figure(figsize=(8,6))
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=y, cmap='coolwarm', edgecolors='k', alpha=0.7)
plt.title("PCA Visualization (First Two Principal Components)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label='Class')
plt.show()
```

## Question 9:KNN Regressor with Distance Metrics and K-Value Analysis
- Task:
1. Generate a synthetic regression dataset
(sklearn.datasets.make_regression(n_samples=500, n_features=10)).
2. Train a KNN regressor with:
a. Euclidean distance (K=5)
b. Manhattan distance (K=5)
c. Compare Mean Squared Error (MSE) for both.
3. Test K=1, 5, 10, 20, 50 and plot K vs. MSE to analyze bias-variance tradeoff.
**Ans:**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error

# 1️⃣ Generate a synthetic regression dataset
X, y = make_regression(n_samples=500, n_features=10, noise=10, random_state=42)

# Split data into training (70%) and testing (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

# 2️⃣ Train KNN regressor with:
# a) Euclidean distance (default metric)
knn_euclidean = KNeighborsRegressor(n_neighbors=5, metric='euclidean')
knn_euclidean.fit(X_train, y_train)
y_pred_euclidean = knn_euclidean.predict(X_test)
mse_euclidean = mean_squared_error(y_test, y_pred_euclidean)

# b) Manhattan distance
knn_manhattan = KNeighborsRegressor(n_neighbors=5, metric='manhattan')
knn_manhattan.fit(X_train, y_train)
y_pred_manhattan = knn_manhattan.predict(X_test)
mse_manhattan = mean_squared_error(y_test, y_pred_manhattan)

print("\n--- Distance Metric Comparison (K=5) ---")
print(f"MSE (Euclidean): {mse_euclidean:.4f}")
print(f"MSE (Manhattan): {mse_manhattan:.4f}")

# 3️⃣ Test different K values to analyze bias-variance tradeoff
k_values = [1, 5, 10, 20, 50]
mse_values = []

for k in k_values:
    knn = KNeighborsRegressor(n_neighbors=k, metric='euclidean')
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mse_values.append(mse)

# Plot K vs. MSE
plt.figure(figsize=(8,5))
plt.plot(k_values, mse_values, marker='o', linestyle='-', color='b')
plt.title("KNN Regressor: K vs. MSE (Bias-Variance Tradeoff)")
plt.xlabel("Number of Neighbors (K)")
plt.ylabel("Mean Squared Error")
plt.grid(True)
plt.show()
```

## Question 10: KNN with KD-Tree/Ball Tree, Imputation, and Real-World Data
- Task:
1. Load the Pima Indians Diabetes dataset (contains missing values).
2. Use KNN Imputation (sklearn.impute.KNNImputer) to fill missing values.
3. Train KNN using:
a. Brute-force method
b. KD-Tree
c. Ball Tree
4. Compare their training time and accuracy.
5. Plot the decision boundary for the best-performing method (use 2 most important
features).
Dataset: Pima Indians Diabetes
**Ans:**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1️⃣ Load the Pima Indians Diabetes dataset
# (Dataset from UCI or Kaggle — available as CSV)
# You can download from: https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv

# Column names as per dataset description
columns = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 
           'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']

# Load dataset
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
data = pd.read_csv(url, names=columns)

print("Dataset shape:", data.shape)
print("Missing values before imputation:\n", (data == 0).sum())

# Replace zeros with NaN for columns where zero is not possible
cols_with_missing = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
data[cols_with_missing] = data[cols_with_missing].replace(0, np.nan)

print("\nMissing values after marking zeros:\n", data.isna().sum())

# 2️⃣ Use KNN Imputation to fill missing values
imputer = KNNImputer(n_neighbors=5)
data_imputed = pd.DataFrame(imputer.fit_transform(data), columns=columns)

print("\nMissing values after KNN imputation:\n", data_imputed.isna().sum())

# Features and target
X = data_imputed.drop('Outcome', axis=1)
y = data_imputed['Outcome']

# Split dataset (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Scale the data for KNN
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 3️⃣ Train KNN using different algorithms
algorithms = ['brute', 'kd_tree', 'ball_tree']
results = {}

for algo in algorithms:
    start = time.time()
    knn = KNeighborsClassifier(n_neighbors=5, algorithm=algo)
    knn.fit(X_train_scaled, y_train)
    y_pred = knn.predict(X_test_scaled)
    end = time.time()
    
    acc = accuracy_score(y_test, y_pred)
    train_time = end - start
    
    results[algo] = {'Accuracy': acc, 'Time': train_time}

# 4️⃣ Compare training time and accuracy
results_df = pd.DataFrame(results).T
print("\n--- KNN Algorithm Comparison ---")
print(results_df)

# Find best performing method
best_algo = results_df['Accuracy'].idxmax()
print(f"\nBest Performing Algorithm: {best_algo}")

# 5️⃣ Plot decision boundary using 2 most important features (Glucose, BMI)
features = ['Glucose', 'BMI']
X2 = data_imputed[features]
y2 = data_imputed['Outcome']

# Scale the 2D features
X2_scaled = scaler.fit_transform(X2)

# Train using the best algorithm
knn_best = KNeighborsClassifier(n_neighbors=5, algorithm=best_algo)
knn_best.fit(X2_scaled, y2)

# Create meshgrid for decision boundary
x_min, x_max = X2_scaled[:, 0].min() - 1, X2_scaled[:, 0].max() + 1
y_min, y_max = X2_scaled[:, 1].min() - 1, X2_scaled[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),
                     np.arange(y_min, y_max, 0.05))

Z = knn_best.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

# Plot the decision boundary
plt.figure(figsize=(8,6))
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')
sns.scatterplot(x=X2_scaled[:,0], y=X2_scaled[:,1], hue=y2, palette='coolwarm', edgecolor='k', s=40)
plt.title(f"Decision Boundary using {best_algo.upper()} (Features: Glucose vs BMI)")
plt.xlabel('Glucose (scaled)')
plt.ylabel('BMI (scaled)')
plt.show()
```
