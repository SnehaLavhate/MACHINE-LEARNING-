## Question 1 : What is Information Gain, and how is it used in Decision Trees?

## Answer:

  * Information Gain (IG) is a measure used in Decision Trees to decide which feature to split on at each step of building the tree. It tells us how much "information" (or reduction in uncertainty) we gain about the target variable after splitting the data based on a particular feature.

#### How it’s used in Decision Trees:

* For each feature, the algorithm:

    - Calculates the Information Gain resulting from splitting the dataset on that feature.
      

* The feature with the highest Information Gain is chosen for the split.

* This process is repeated recursively to grow the tree until stopping criteria are met (e.g., max depth or pure leaves).

## Question 2: What is the difference between Gini Impurity and Entropy?
* Hint: Directly compares the two main impurity measures, highlighting strengths,
weaknesses, and appropriate use cases.

**Answer:**

#### Key differences between Gini Impurity and Entropy in point form:


1. **Definition:**

   * *Gini Impurity* measures how often a randomly chosen element would be incorrectly classified.
   * *Entropy* measures the amount of uncertainty or disorder in the data.

2. **Formula:**

   * Gini: ( G = 1 - \sum p_i^2 )
   * Entropy: ( H = -\sum p_i \log_2(p_i) )

3. **Range:**

   * Gini ranges from **0 to 0.5** (for two classes).
   * Entropy ranges from **0 to 1** (for two classes).

4. **Computation:**

   * Gini is **faster** to compute since it doesn’t involve logarithms.
   * Entropy is **slower** due to the log base 2 calculation.

5. **Bias:**

   * Gini is slightly **biased toward features with more distinct values**.
   * Entropy is **less biased** in that regard.

6. **Algorithm Use:**

   * Gini is used in **CART (Classification and Regression Trees)**.
   * Entropy is used in **ID3** and **C4.5** algorithms.

7. **Interpretation:**

   * Gini focuses on **minimizing misclassification**.
   * Entropy focuses on **maximizing information gain**.

8. **Practical Choice:**

   * Both usually give similar results, but **Gini** is often preferred in practice because it’s simpler and faster to compute.


## Question 3:What is Pre-Pruning in Decision Trees?

**Answer:**

**Pre-pruning** (also called **Early Stopping**) is a technique used in **Decision Trees** to **stop the tree from growing too large** during its construction — before it overfits the data.


### 🔹 **Explanation:**

Instead of allowing the tree to grow fully and then trimming it (as in post-pruning), pre-pruning **prevents further splitting** of a node if certain conditions are met.


### 🔹 **Common Stopping Criteria for Pre-Pruning:**

1. **Maximum Depth:** Stop splitting when the tree reaches a specified depth.
2. **Minimum Samples per Split:** Stop if a node has fewer samples than a set threshold.
3. **Minimum Information Gain / Gini Reduction:** Stop if the improvement in purity is too small.
4. **Maximum Number of Nodes or Leaves:** Limit the overall size of the tree.
5. **Validation Accuracy Check:** Stop if further splitting doesn’t improve accuracy on validation data.


### 🔹 **Advantages:**

* Prevents **overfitting** early.
* Reduces **training time**.
* Keeps the tree **simpler and more interpretable**.


### 🔹 **Disadvantages:**

* Might **underfit** if stopped too early, missing important splits.



````python
## Question 4:Write a Python program to train a Decision Tree Classifier using Gini Impurity as the criterion and print the feature importances (practical).

# Hint: Use criterion='gini' in DecisionTreeClassifier and access .feature_importances_.

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

# Load sample dataset (Iris dataset)
data = load_iris()
X = data.data          # Features
y = data.target        # Target labels

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
````

````python
# Create a Decision Tree Classifier using Gini Impurity
clf = DecisionTreeClassifier(criterion='gini', random_state=42)


clf.fit(X_train, y_train)   # Train the model

# Print feature importances
print("Feature Importances:")
for feature_name, importance in zip(data.feature_names, clf.feature_importances_):
    print(f"{feature_name}: {importance:.4f}")

# Optional: Check model accuracy
accuracy = clf.score(X_test, y_test)
print("\nModel Accuracy:", accuracy)
````

## Question 5: What is a Support Vector Machine (SVM)?

**Answer:**

A **Support Vector Machine (SVM)** is a **supervised machine learning algorithm** used for **classification** and **regression** tasks — most commonly for classification.



### 🔹 **Concept:**

SVM works by finding the **best boundary (called a hyperplane)** that separates data points of different classes **with the maximum margin** (i.e., the largest possible distance between the boundary and the nearest data points of any class).
These closest points are called **Support Vectors**.



### 🔹 **Key Ideas:**

1. **Hyperplane:**
   The decision boundary that separates different classes.
2. **Margin:**
   The distance between the hyperplane and the nearest data points. SVM tries to **maximize** this margin.
3. **Support Vectors:**
   The data points that lie closest to the hyperplane; they determine the position and orientation of the boundary.
4. **Kernel Trick:**
   Used to handle **non-linear data** by mapping it into a higher-dimensional space (e.g., using linear, polynomial, or RBF kernels).



### 🔹 **Advantages:**

* Works well with **high-dimensional data**.
* Effective when there is a **clear margin of separation**.
* Uses only support vectors, so it’s **memory efficient**.



### 🔹 **Disadvantages:**

* **Computationally expensive** for large datasets.
* Performance depends on choosing the **right kernel and parameters**.


## Question 6: What is the Kernel Trick in SVM?

**Answer:**

The **Kernel Trick** is a technique used in **Support Vector Machines (SVM)** to handle **non-linear data** by mapping it into a **higher-dimensional space** where it becomes easier to separate the classes with a linear boundary.



### **Explanation:**

Some datasets cannot be separated by a straight line in their original space.
The kernel trick allows SVM to perform this transformation **implicitly**, without actually computing the coordinates in the higher dimension.

Instead of transforming the data using a mapping function `φ(x)`,
SVM uses a **kernel function** `K(x_i, x_j)` that computes the **dot product** in the higher-dimensional space directly:

```
K(x_i, x_j) = φ(x_i) · φ(x_j)
```

This means we don’t need to calculate φ(x) explicitly — the kernel function does it internally.

### **Common Kernel Functions:**

1. **Linear Kernel:**

   ```
   K(x_i, x_j) = x_i · x_j
   ```

2. **Polynomial Kernel:**

   ```
   K(x_i, x_j) = (x_i · x_j + c)^d
   ```

3. **RBF (Radial Basis Function) / Gaussian Kernel:**

   ```
   K(x_i, x_j) = exp(-γ * ||x_i - x_j||^2)
   ```


### **Advantages:**

* Handles **non-linear relationships** efficiently.
* Avoids the need to compute high-dimensional transformations explicitly.



````python
## Question 7: Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies.
# Hint:Use SVC(kernel='linear') and SVC(kernel='rbf'), then compare accuracy scores after fitting on the same dataset.
    
# Import necessary libraries
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load the Wine dataset
data = load_wine()
X = data.data
y = data.target

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create and train SVM with Linear Kernel
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train, y_train)

# Create and train SVM with RBF Kernel
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train, y_train)

# Make predictions
y_pred_linear = svm_linear.predict(X_test)
y_pred_rbf = svm_rbf.predict(X_test)

# Calculate accuracies
acc_linear = accuracy_score(y_test, y_pred_linear)
acc_rbf = accuracy_score(y_test, y_pred_rbf)

print("Accuracy with Linear Kernel:", acc_linear)
print("Accuracy with RBF Kernel:", acc_rbf)

# Compare which is better
if acc_linear > acc_rbf:
    print("\nLinear kernel performed better.")
elif acc_rbf > acc_linear:
    print("\nRBF kernel performed better.")
else:
    print("\nBoth kernels performed equally well.")
````

## Question 8: What is the Naïve Bayes classifier, and why is it called "Naïve"?

**Answer:**

The **Naïve Bayes classifier** is a **probabilistic machine learning algorithm** based on **Bayes’ Theorem**, used mainly for **classification tasks** (like spam detection, sentiment analysis, etc.).



### 🔹 **Bayes’ Theorem:**

```
P(A|B) = (P(B|A) * P(A)) / P(B)
```

Where:

* `P(A|B)` = Probability of class A given data B (posterior)
* `P(B|A)` = Probability of data B given class A (likelihood)
* `P(A)` = Probability of class A (prior)
* `P(B)` = Probability of the data (evidence)



### 🔹 **How Naïve Bayes Works:**

1. Calculates the **prior probability** for each class.
2. Uses the **likelihood** of the features for each class.
3. Applies **Bayes’ Theorem** to compute the **posterior probability** for each class.
4. The class with the **highest posterior probability** is chosen as the prediction.



### 🔹 **Why it is called “Naïve”:**

It’s called *Naïve* because it **assumes that all features are independent of each other** given the class label — an assumption that is rarely true in real-world data.
Despite this unrealistic assumption, it works surprisingly well in many practical applications.



### 🔹 **Advantages:**

* Simple and fast to train.
* Works well with **high-dimensional** data (like text).
* Performs well even with small datasets.



## Question 9: Explain the differences between Gaussian Naïve Bayes, Multinomial Naive Bayes, and Bernoulli Naïve Bayes

**Answer:**

The **Naïve Bayes classifier** has several variants based on the type of data it works best with. Here’s the difference between **Gaussian**, **Multinomial**, and **Bernoulli Naïve Bayes**:


### 1. **Gaussian Naïve Bayes (GNB):**

* **Data Type:** Continuous/numerical features.
* **Assumption:** Each feature follows a **normal (Gaussian) distribution**.
* **Example Use Case:** Predicting species of iris flowers based on sepal/petal measurements.
* **Likelihood Formula:**

  ```
  P(x_i | y) = (1 / sqrt(2πσ_y^2)) * exp(-(x_i - μ_y)^2 / (2σ_y^2))
  ```

  Where μ and σ are mean and standard deviation of the feature for class y.



### 2. **Multinomial Naïve Bayes (MNB):**

* **Data Type:** Count-based or discrete features (like word counts in text).
* **Assumption:** Features represent **frequency or counts** of events.
* **Example Use Case:** Text classification, spam detection (based on word counts).
* **Key Idea:** Uses the **multinomial distribution** to calculate likelihood of features.



### 3. **Bernoulli Naïve Bayes (BNB):**

* **Data Type:** Binary/boolean features (0 or 1).
* **Assumption:** Features are **present or absent**.
* **Example Use Case:** Text classification with **binary occurrence** of words (word present = 1, absent = 0).
* **Key Idea:** Uses the **Bernoulli distribution** to calculate likelihood of features.



**In short:**

* **Gaussian** → numerical features, assumes normal distribution.
* **Multinomial** → count-based features, uses multinomial distribution.
* **Bernoulli** → binary features, uses Bernoulli distribution.
